```python
import time
import random
import sys

def interpolation_search(arr, target):
    """
    Interpolation Search Algorithm
    Time Complexity: O(log log n) average, O(n) worst case
    Space Complexity: O(1)
    """
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1
        if low == high: # If only one element left
            if arr[low] == target:
                return low, comparisons
            else: # If the single element is not the target
                return -1, comparisons

        # Handle division by zero if all elements in the range are the same
        # This occurs if arr[high] == arr[low] but target is not found in the 'if low == high' block above
        if arr[high] == arr[low]:
            # If target was not found in the previous 'if low == high' block,
            # and arr[high] == arr[low], it means target is not in this range.
            return -1, comparisons

        # Interpolation formula
        pos = low + int(((target - arr[low]) * (high - low)) /
                        (arr[high] - arr[low]))

        if arr[pos] == target:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else: # arr[pos] > target
            high = pos - 1
    return -1, comparisons # Target not found in the array or range exhausted

def binary_search(arr, target):
    """Binary Search for comparison"""
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, comparisons
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

def performance_analysis():
    sizes = [1000, 5000, 10000, 50000, 100000]
    print(f"{'Size':>10} {'IS Time(ms)':>14} {'BS Time(ms)':>14} "
          f"{'IS Comparisons':>16} {'BS Comparisons':>16}")
    print('-' * 75)

    for size in sizes:
        # Generate a sorted array with unique elements
        arr = sorted(random.sample(range(size * 10), size))
        target = arr[random.randint(0, size - 1)] # Target is guaranteed to be in array

        # Interpolation Search timing
        start = time.perf_counter()
        for _ in range(100): # Run multiple times for better average
            idx_is, comp_is = interpolation_search(arr, target)
        is_time = (time.perf_counter() - start) / 100 * 1000 # Convert to milliseconds

        # Binary Search timing
        start = time.perf_counter()
        for _ in range(100): # Run multiple times for better average
            idx_bs, comp_bs = binary_search(arr, target)
        bs_time = (time.perf_counter() - start) / 100 * 1000 # Convert to milliseconds

        print(f"{size:>10} {is_time:>14.4f} {bs_time:>14.4f} "
              f"{comp_is:>16} {comp_bs:>16}")

# --- Main ---
arr = [2, 5, 10, 15, 23, 35, 48, 60, 75, 90, 105, 120]
target = 35
idx, comps = interpolation_search(arr, target)
print(f"Array: {arr}")
print(f"Searching for: {target}")
print(f"Found at index: {idx}, Comparisons: {comps}")
print()
performance_analysis()
```
